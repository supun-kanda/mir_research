```python
import librosa
```

```python
filename = librosa.example('nutcracker')
```

```python
y, sr = librosa.load(filename)

# print("y", y)
# print("sr", sr)
```

```python
tempo, beat_frames = librosa.beat.beat_track(y=y, sr=sr)
```
